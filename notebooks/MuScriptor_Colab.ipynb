### Notes

- **Dynamic velocity** estimates MIDI velocity from the source audio and is enabled by default.
- **Best effort tempo** keeps a useful average BPM for coherent live performances instead of unnecessarily falling back to 120 BPM.
- **Quantize** is off by default so expressive timing is preserved.
- Changing the model/device/dtype unloads the previous checkpoint before loading the new one.
