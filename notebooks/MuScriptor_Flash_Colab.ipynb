# MuScripter Flash · Colab

Run **MuScripter Flash** from Google Colab with realtime MIDI output back to your PC. The browser streams microphone audio to Colab at roughly **32 ms cadence**; FlashEngine detects notes and the browser forwards NoteOn/NoteOff through **Web MIDI** to loopMIDI, a hardware MIDI port, or your DAW.

> Colab adds browser/network latency, so the native ≤250 ms target is not an end-to-end guarantee here. The UI shows Flash engine latency separately from rough transport/queue lag.

A GPU is **not required** for the current Flash spectral MVP. Desktop Chrome or Edge is recommended for Web MIDI.

## Launch Flash

1. On Windows, create/start a loopMIDI port (for example `MuScripter Flash`).

3. Enter the MIDI output name, press **Connect realtime MIDI**, and allow MIDI access.
4. Start microphone recording. Detected notes are sent to the local MIDI port immediately.
5. Use **Panic / All Notes Off** if needed; stopping recording also flushes active notes and creates a `.mid` file.
